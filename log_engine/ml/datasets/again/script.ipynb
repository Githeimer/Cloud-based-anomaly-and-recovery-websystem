```python
import os

# List of the input jsonl files shown in the dataset
input_files = [
    "healthy_interleaved_2k.jsonl",
    "brute_force_5k_windowed.jsonl",
    "refresh_abuse_5k_windowed.jsonl"
]
output_file = "logs_combined.jsonl"

def combine_jsonl_files(files, output_path):
    print(f"Starting to combine files into {output_path}...")
    
    total_lines = 0
    # Open the output file in write mode
    with open(output_path, 'w', encoding='utf-8') as outfile:
        for file_name in files:
            if os.path.exists(file_name):
                print(f"Processing: {file_name}...")
                line_count = 0
                
                # Read line-by-line to be memory efficient
                with open(file_name, 'r', encoding='utf-8') as infile:
                    for line in infile:
                        # Strip accidental empty lines but keep valid JSON lines
                        if line.strip():
                            outfile.write(line.strip() + '\n')
                            line_count += 1
                
                print(f"-> Successfully added {line_count} lines from {file_name}")
                total_lines += line_count
            else:
                print(f"Warning: {file_name} not found in the current directory.")

    print(f"\nDone! Combined {total_lines} total lines into {output_path}")

if __name__ == "__main__":
    combine_jsonl_files(input_files, output_file)
```
